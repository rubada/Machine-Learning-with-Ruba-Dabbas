#### **a.2 Categorical Data:**
Categorical data refers to the data that is grouped into distinct categories rather than measured numerically. It is often used to classify observations based on labels or names rather than numerical values. There are two main types:
- Nominal Data: Categories have no inherent order (e.g., colors, gender, country names).
- Ordinal Data: Categories have a meaningful order (e.g., in our dataset, grade, condition, and they are numeric data, they may be non-numeric data such as (A > B > C > .....).

```python
# You can check the size of each attribute in this Ordinal Data:
grade_count = df_houses.groupby("grade").size()

print(grade_count)
```

```python
grade_count = df_houses.groupby("condition").size()

print(grade_count)
```

```python
# To change the "condition" column values from numbers to strings,
# use the "replace" function:
df_houses["condition_cat"] = df_houses["condition"].replace({1: "E", 2: "D", 3: "C", 4: "B", 5: "A"})

df_houses["condition_cat"].unique()
```

```python
# As for the grade column, it represents the overall construction
# and design quality of the house, as assessed by King County’s
# grading system:
# - Low Quality: 1–4
# - Average Quality: 5–7
# - Good Quality: 8–10
# - Luxury: 11–13

# OR
# To change the "grade" column values from numbers to strings,
# use the "map" function:
mapping = {
    1: "Low", 2: "Low", 3: "Low", 4: "Low",
    5: "Average", 6: "Average", 7: "Average",
    8: "Good", 9: "Good", 10: "Good",
    11: "Luxury", 12: "Luxury", 13: "Luxury",
}
df_houses["grade_cat"] = df_houses["grade"].map(mapping)

df_houses["grade_cat"].unique()
```

```python
houses_cat = df_houses[['condition_cat', 'grade_cat']]
houses_cat.describe(include=['O'])
```

- count → total non-null entries
- unique → number of distinct categories
- top → the most common category
- freq → how many times that top category appears

This table is useful because it gives you a quick diagnostic snapshot of your categorical data:
- **Data completeness** → "count" shows how many non-null values exist per column. You can spot missing data patterns immediately.
  
- **Category richness** → "unique" tells you how many distinct categories exist. This helps you determine whether a feature is diverse (with many categories) or narrow (with few categories).

- **Dominant category** → "top "and "freq" highlight the most common value and how often it appears. This reveals an imbalance or skew in categorical features.

- **Feature usefulness** → If one category dominates (high freq relative to count), the feature may have low predictive power and could be dropped or transformed.

- **Data validation** → You can catch anomalies, like placeholder values (“Unknown”, “NA”) showing up as the top category, which signals data quality issues.

- **Encoding strategy** → Knowing the top category helps decide between one-hot encoding, target encoding (mean encoding), or grouping rare categories. For example, if freq is high, you might collapse rare categories into “Other.”

#### **a.3 Correlations Between Attributes:**
In machine learning, correlation between attributes is significant because it helps in understanding relationships between features, improving model performance, and avoiding redundancy.

Correlation is the relationship between two variables and how they may or may not change together.
<br>Positive correlations between attributes indicate that both increase together, while negative ones mean one decreases as the other increases.
<br>A correlation of -1 or 1 shows a strong correlation between the attributes. Whereas a value of 0 shows no correlation at all.

Some machine learning algorithms, such as linear and logistic regression, may perform poorly if the correlation between attributes is strong.
<br>Thus, it is a good idea to review all of the pairwise correlations of the attributes in your dataset.

The most common method for calculating correlation is Pearson’s Correlation Coefficient, which assumes a normal distribution of the attributes involved.

We will use the "corr" method, which is only applicable to a numeric dataset.
<br>If the dataset has non-numeric attributes, remove them when using the "corr" method.
<br> In our dataset, some columns shouldn't be used with the "corr" method:
- The "date" is non-numeric.
- The "id " column is just an ID for each house.
- The categorical data, the "grade" and "condtion" columns.

```python
# Dropping unnecessary columns:
data_drop_non_num = df_houses.drop(["date", "id", "grade", "condition", "condition_cat", "grade_cat"], axis=1)

data_corr = data_drop_non_num.corr(method="pearson")
display(data_corr)
```

Correlation helps us with:
- Feature Selection: Highly correlated features can introduce redundancy, so removing one can simplify the model without losing valuable information.
- Multicollinearity Detection: In regression models, strong correlations between independent variables can distort predictions, making it harder to interpret coefficients.
- Predictive Power: Some features may have strong correlations with the target variable, making them more useful for predictions.

We can use the "spearman" correlation with oridnal data:

```python
data_corr = df_houses.loc[:, ["grade", "condition"]].corr(method="spearman")
display(data_corr)
```
